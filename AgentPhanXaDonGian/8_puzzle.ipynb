```python
import time

START = [1, 2, 3, 
         4, 0, 6, 
         7, 5, 8]

GOAL  = [1, 2, 3, 
         4, 5, 6, 
         7, 8, 0]
```

```python
def print_board(state):
    for i in range(0, 9, 3):
        print(f"| {state[i]} {state[i+1]} {state[i+2]} |")
    print("-" * 12)
```

```python
class ModelBasedReflexAgent:
    def __init__(self, goal):
        self.goal = goal
        self.model = set()

    def think(self, state):
        z = state.index(0)
        r, c = divmod(z, 3)
        
        possible_moves = []
        directions = {
            (-1, 0): "Lên", 
            (1, 0): "Xuống", 
            (0, -1): "Trái", 
            (0, 1): "Phải"
        }

        print(f"\n[AI THOUGHT]: Đang ở vị trí ô trống ({r}, {c}).")
        
        choices = []
        for (dr, dc), name in directions.items():
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_state = list(state)
                target_idx = nr * 3 + nc
                new_state[z], new_state[target_idx] = new_state[target_idx], new_state[z]
                
                state_tuple = tuple(new_state)
                if state_tuple not in self.model:
                    score = self.evaluate(new_state)
                    choices.append((target_idx, name, score, state_tuple))
        
        choices.sort(key=lambda x: x[2])
        
        if not choices:
            print("[AI THOUGHT]: Không còn nước đi mới! Đang bị kẹt.")
            return None
            
        best_move = choices[0]
        print(f"[AI THOUGHT]: Tìm thấy {len(choices)} hướng đi mới. Chọn hướng '{best_move[1]}' vì tối ưu nhất.")
        return best_move

    def evaluate(self, state):
        return sum(1 for i in range(9) if state[i] != self.goal[i] and state[i] != 0)

    def act(self, state):
        move_info = self.think(state)
        if move_info:
            target_idx, name, score, state_tuple = move_info
            self.model.add(state_tuple)
            return target_idx
        return None
```

```python
def run_simulation():
    current_state = list(START)
    agent = ModelBasedReflexAgent(GOAL)
    
    agent.model.add(tuple(START))
    
    print("=== TRẠNG THÁI BẮT ĐẦU ===")
    print_board(current_state)
    
    step = 0
    while current_state != GOAL and step < 50:
        step += 1
        time.sleep(0.5)
        
        move_idx = agent.act(current_state)
        
        if move_idx is None:
            break
            
        z_idx = current_state.index(0)
        current_state[z_idx], current_state[move_idx] = current_state[move_idx], current_state[z_idx]
        
        print(f"--- BƯỚC {step} ---")
        print_board(current_state)

    if current_state == GOAL:
        print("KẾT QUẢ: AI đã đưa ma trận về đúng thứ tự thành công!")
    else:
        print("KẾT QUẢ: Thất bại hoặc quá giới hạn bước đi.")
```

```python
if __name__ == "__main__":
    run_simulation()
```
