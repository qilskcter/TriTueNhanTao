# 8-Puzzle BFS

```python
start_state = (1, 2, 3, 
               4, 5, 6, 
               7, 0, 8)

goal_state = (1, 2, 3, 
              4, 5, 6, 
              7, 8, 0)
```

```python
def find_blank(state):
    for i in range(9):
        if state[i] == 0:
            return i

def get_next_states(state):
    next_states = []
    blank_index = find_blank(state)
    row = blank_index // 3
    col = blank_index % 3

    moves = []
    if row > 0:
        moves.append(blank_index - 3)
    if row < 2:
        moves.append(blank_index + 3)
    if col > 0:
        moves.append(blank_index - 1)
    if col < 2:
        moves.append(blank_index + 1)

    for next_index in moves:
        temp_list = list(state)
        temp_list[blank_index], temp_list[next_index] = temp_list[next_index], temp_list[blank_index]
        next_states.append(tuple(temp_list))
        
    return next_states

def print_board(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print(f"\n")
```

```python
def solve_8_puzzle(start, goal):
    queue = [[start]]
    visited = set([start])

    while len(queue) > 0:
        path = queue.pop(0)
        current_state = path[-1]

        if current_state == goal:
            return path

        for next_state in get_next_states(current_state):
            if next_state not in visited:
                visited.add(next_state)
                new_path = list(path)
                new_path.append(next_state)
                queue.append(new_path)
                
    return None
```

```python
result = solve_8_puzzle(start_state, goal_state)

if result:
    print("Tim thay loi giai!")
    print("So buoc di:", len(result) - 1)
    print()
    for step in result:
        print_board(step)
else:
    print("Khong tim thay loi giai.")
```
